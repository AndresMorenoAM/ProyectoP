```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 5, 20) #inicio, #fin, #catidad de datos
y = x**2

plt.plot(x,y, color = "purple", marker = "H", linestyle = "--", linewidth = 2)
plt.title("$f(x)=x^2$")
plt.ylabel("$x^2$")
plt.xlabel("$x$")
plt.grid()
plt.savefig("Funcioncuadradada.png")
plt.show()
```

```python
plt.pie(y)
plt.show()
```

```python
plt.scatter(x,y)
plt.show()
```

```python
plt.boxplot(y)
plt.show()
```

```python
x = np.linspace(0,40,500)
y = np.sin(x)
plt.plot(x,y)
plt.title("$F(x) = Sen(x)$")
plt.xlabel("$x$")
plt.ylabel("$Sen(x)$")
plt.xlim(0, 30) #dominio
plt.ylim(-2,2) #rango
plt.grid()
plt.show()
```

```python
x = np.linspace(0,10,100)
y =  np.sin(x)
y2 = np.cos(x)


plt.plot(x,y)
plt.plot(x,y2)
plt.title("$Sen(x), Cos(x)$")
plt.xlabel("x")
plt.legend(["$Sen(x)$","$Cos(x)$"])

plt.grid()
```

```python
from matplotlib import legend
#Subplots

plt.subplot( 2, 1, 1) #filas, #columnas,
plt.plot(x,y, label="$Sen(x)$")
plt.title("$Sen(x)$")
plt.xlabel("$x$")
plt.legend()
plt.grid()



plt.subplot(2, 1, 2)
plt.plot(x,y2, color = "red", label="$Cos(x)$")
plt.title("$Cos(x)$")
plt.xlabel("$x$")
plt.legend()
plt.grid()

plt.tight_layout()
```

# Tarea


```python
x = np.random.rand(30) * 5
y = x**2 + np.random.normal(0, 2, 30)

plt.scatter(x, y,color="darkgreen",marker="o",s=50,alpha=0.7)

plt.title(r"Gráfico de Dispersión: $y = x^2 + \epsilon$")
plt.xlabel(r"$Variable \; X$")
plt.ylabel(r"$Variable \; Y$")

plt.grid(True)
plt.savefig("GraficodeDispersion.png")
plt.show()
```

```python
categorias = ["A", "B", "C", "D", "E"]
valores = [12, 25, 18, 30, 15]

plt.bar(categorias,valores,color="navy",edgecolor="black",linewidth=1.2,alpha=0.8)

plt.title("Distribución por Categorías")
plt.xlabel("$Categorías$")
plt.ylabel("$Frecuencia$")

plt.grid(axis="y", linestyle="--")
plt.savefig("GraficoBarras.png")
plt.show()
```
